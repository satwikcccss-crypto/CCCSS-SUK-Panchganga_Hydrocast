# HydroCast: Offline Physical Emulator Tutorial

This notebook demonstrates how to load the `hms.runner` module locally and simulate a flood wave without running the entire real-time pipeline.

```python
import sys
from pathlib import Path
from datetime import datetime, timezone

# Ensure the src directory is in the python path
sys.path.append(str(Path.cwd().parent))

from src.hms.runner import execute_hec_hms
import numpy as np
```

```python
# Generate a synthetic 90-hour storm (heavy monsoon simulation)
synthetic_hyetograph = {}
for sid in ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9"]:
    rain = np.zeros(90, dtype=np.float32)
    # 10mm/hr of rain for 24 hours starting at hour 10
    rain[10:34] = 10.0
    synthetic_hyetograph[sid] = rain
```

```python
run_dt = datetime(2026, 9, 28, 6, 0, tzinfo=timezone.utc)

# Execute the physics emulator!
results = execute_hec_hms(
    run_dt=run_dt,
    subbasin_hyetographs=synthetic_hyetograph,
    live_stage_m=531.0, # Current river height
)

print(f"Peak Discharge: {results['peak_discharge_m3s']} m³/s")
print(f"Peak Lead Time: {results['lead_hours_to_peak']} hours")
print(f"Soil Moisture Condition: {results['amc']}")
```
